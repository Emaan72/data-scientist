## Write a program to calculate the sum of numbers from 1 to N.
### Example:

* Enter number: 5
* Output: 15

```python
n = int(input("Enter number: "))
total = 0

for i in range(1, n + 1):
    total = total + i

print("Output:", total)
```

## Write a program to count how many digits exist in a number using a loop.
### Example:

* Input: 45892
* Output: 5

```python
n = int(input("Enter number: "))
count = 0

while n > 0:
    n = n // 10
    count += 1

print("Output:", count)
```

## Write a program to calculate the factorial of a number.
### Example:

* Input : 5
* Output 120

```python
n = int(input("Enter number: "))
fact = 1

for i in range(1, n + 1):
    fact = fact * i

print("Output:", fact)
```

## Write a program to calculate:

* base ^ exponent
* without using **

```python
base = int(input("Enter base: "))
exponent = int(input("Enter exponent: "))
result = 1

for i in range(exponent):
    result = result * base

print("Output:", result)
```

## Write a program to count vowels in a string.
### Example:
* Input: pakistan
* Output: 3

```python
s = input("Enter a string: ")
count = 0

for ch in s:
    if ch in "aeiouAEIOU":
        count += 1

print("Output:", count)
```

## Write a program to find the maximum value in a list without using max().

```python
numbers = [10, 25, 7, 40, 15]
maximum = numbers[0]

for num in numbers:
    if num > maximum:
        maximum = num

print("Maximum value:", maximum)
```

## 🏪 1. Shopping Cart System

### Goal:
    Create a shopping cart program that allows users to add items to their cart.
    Stop adding items when the user types "done".

### Steps:

    Ask the user to enter product names.
    
    Add each item to a list.
    
    If user types "done" → use break.
    
    Print all items added to the cart.

### Example:

    Enter item: Shirt  
    Enter item: Shoes  
    Enter item: done  
    Items in your cart: ['Shirt', 'Shoes']

```python
cart = []

while True:
    item = input("Enter item: ")

    if item == "done":
        break

    cart.append(item)

print("Items in your cart:", cart)
```
